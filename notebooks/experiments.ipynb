```python
# Compare experiment runs logged by src/train.py
runs = []
for path in sorted(glob.glob('../results/run_*.json')):
    with open(path) as f:
        runs.append(json.load(f))

for r in runs:
    a = r['args']
    print(a['model'], '| pretrained=', a['pretrained'], '| augment=', a['augment'],
          '| lr=', a['lr'], '-> best_val_f1=', round(r['best_val_f1'], 4))
```
